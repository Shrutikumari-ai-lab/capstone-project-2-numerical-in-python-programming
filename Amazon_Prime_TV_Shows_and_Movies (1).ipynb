## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#Step 1: Create Copies of Original Datasets
titles = titles_df.copy()
credits = credits_df.copy()
````

````python
#Step 2: Check Duplicate
# Check duplicate rows

print("Duplicate rows in Titles Dataset :", titles.duplicated().sum())
print("Duplicate rows in Credits Dataset :", credits.duplicated().sum())
````

````python
#Step 3: Remove Duplicate Records
# Remove duplicate rows

titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)

print("Duplicates removed successfully.")
````

````python
#Step 4: Check Missing Values
# Check missing values

print("Titles Dataset Missing Values")
print(titles.isnull().sum())

print("\nCredits Dataset Missing Values")
print(credits.isnull().sum())
````

````python
#Step 5: Handle Missing Values (Titles Dataset)
# Fill numerical columns with median

titles['imdb_score'].fillna(titles['imdb_score'].median(), inplace=True)
titles['tmdb_score'].fillna(titles['tmdb_score'].median(), inplace=True)
titles['tmdb_popularity'].fillna(titles['tmdb_popularity'].median(), inplace=True)

# Fill categorical column with mode

titles['age_certification'].fillna(
    titles['age_certification'].mode()[0],
    inplace=True
)

# Replace missing seasons with 0
titles['seasons'].fillna(0, inplace=True)
````

````python
#Step 6: Handle Missing Values (Credits Dataset)
# Replace missing character names

credits['character'].fillna("Unknown", inplace=True)
````

````python
#Step 7: Convert Data Types
# Convert release year to integer

titles['release_year'] = titles['release_year'].astype(int)

# Convert runtime to numeric

titles['runtime'] = pd.to_numeric(
    titles['runtime'],
    errors='coerce'
)
````

````python
#Step 8: Verify Missing Values After Cleaning
print("Titles Dataset")
print(titles.isnull().sum())

print("\nCredits Dataset")
print(credits.isnull().sum())
````

````python
#Step 9: Verify Dataset Shape
print("Titles Dataset Shape :", titles.shape)
print("Credits Dataset Shape :", credits.shape)
````

### What all manipulations have you done and insights you found?

Duplicate records were removed from both datasets to improve data quality. Missing values in important columns were handled using appropriate techniques such as median, mode, and 0 for the seasons column, while missing character values were replaced with "Unknown". Data types were also verified and corrected where necessary.

Insights:

Removed 3 duplicate rows from the Titles dataset and 56 from the Credits dataset.
The Seasons and Age Certification columns had the highest missing values.

````python
import ast

# 1.Which are the top 10 countries releasing the most movies with IMDb > 7 and TMDb > 7.5

# filter movies with good IMDb and TMDb scores
good_movies = titles[(titles['imdb_score'] > 7) & (titles['tmdb_score'] > 7.5) & (titles['type'] == 'MOVIE')]

# Clean the production_countries column (string to list)
good_movies['production_countries_list'] = good_movies['production_countries'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

#explode to get one country per row
explode = good_movies.explode("production_countries_list")

# remove empty values
explode = explode[explode['production_countries_list'].str.strip() != ""]

top_countries = (explode.groupby("production_countries_list", as_index=False)["title"].count().rename(columns={"title": "high_rated_movies"}).sort_values(by="high_rated_movies", ascending=False).head(10))

top_countries
````

````python
import ast

# 2. Which are the top 10 most common genres on Amazon Prime, and how many titles belong to each genre?

# Clean 'genres' column (string to list)
titles['genres_list'] = titles['genres'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

# Explode genres into separate rows
explode_genres = titles.explode("genres_list")

# Remove empty values
explode_genres = explode_genres[explode_genres["genres_list"].str.strip() != ""]

# Group by genres and count titles
top_genres = (
    explode_genres.groupby("genres_list", as_index=False)
    ["title"].count()
    .rename(columns={"title": "total_titles"})
    .sort_values(by="total_titles", ascending=False)
    .head(10)
)

top_genres
````

````python
import ast

# 3. Which are the top 10 production countries contributing the most content on Amazon Prime?

# Clean 'production_countries' column (string to list)
titles['production_countries_list'] = titles['production_countries'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

# Explode countries into separate rows
explode_countries = titles.explode("production_countries_list")

# Remove empty values
explode_countries = explode_countries[explode_countries["production_countries_list"].str.strip() != ""]

# Group by countries and count total titles
top_countries_content = (
    explode_countries.groupby("production_countries_list", as_index=False)
    ["title"].count()
    .rename(columns={"title": "total_titles"})
    .sort_values(by="total_titles", ascending=False)
    .head(10)
)

top_countries_content
````

````python
# 4.  How has the number of movies and shows released each year changed over time?

# Group by release year and type (Movie/Show)
yearly_trend = (
    titles.groupby(["release_year", "type"], as_index=False)
    ["title"].count()
    .rename(columns={"title": "total_titles"})
    .sort_values(by="release_year")
)

yearly_trend
````

````python
# 5. Which are the top 10 movies/shows with the highest TMDb popularity score?

# Select top 10 titles with highest TMDb popularity
# Drop duplicate titles so each movie/show appears only once
top_popular = (
    titles.drop_duplicates(subset="title")
    [["title", "type", "tmdb_popularity", "imdb_score", "tmdb_score"]]
    .sort_values(by="tmdb_popularity", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

print(top_popular)
````

# KPIs (Key Performance Indicators) Calculation

# 1. Content Library Health

````python

# KPI 1: Content Library Overview

# Calculate total unique titles available on Amazon Prime
total_titles = titles['title'].nunique()

# Calculate total number of movies
total_movies = (titles['type'] == 'MOVIE').sum()

# Calculate total number of TV shows
total_shows = (titles['type'] == 'SHOW').sum()

# Calculate percentage of movies in the content library
movie_share = (total_movies / len(titles)) * 100

print(" Total Titles:", total_titles)
print(" Total Movies:", total_movies)
print(" Total TV Shows:", total_shows)
print(f" Movie Share: {movie_share:.2f}%")
````

# 2. Audience Satisfaction KPIs

````python

# KPI 2: Audience Satisfaction

# Calculate average IMDb rating
avg_imdb = titles['imdb_score'].mean()

# Calculate average TMDB rating
avg_tmdb = titles['tmdb_score'].mean()

# Filter titles having IMDb rating greater than or equal to 8
high_rated = titles[titles['imdb_score'] >= 8]

# Calculate percentage of highly rated titles
high_rated_percent = (len(high_rated) / len(titles)) * 100

print(" Average IMDb Rating:", round(avg_imdb,2))
print(" Average TMDB Rating:", round(avg_tmdb,2))
print(" High Rated Titles:", len(high_rated))
print(f" High Rated Content: {high_rated_percent:.2f}%")
````

# 3. Global Content Diversity

````python

# KPI 3: Global Content Diversity

# Count unique production countries
total_countries = titles['production_countries'].explode().nunique()

# Count unique genres
total_genres = titles['genres'].explode().nunique()

# Count unique age certifications
total_certifications = titles['age_certification'].nunique()

print(" Production Countries:", total_countries)
print(" Unique Genres:", total_genres)
print(" Age Certifications:", total_certifications)
````

# 4. Growth KPIs

````python

# KPI 4: Content Growth

# Find the earliest release year
earliest_release = titles['release_year'].min()

# Find the latest release year
latest_release = titles['release_year'].max()

# Calculate average release year
average_release = titles['release_year'].mean()

print(" Earliest Release:", earliest_release)
print(" Latest Release:", latest_release)
print(" Average Release Year:", round(average_release))
````

# 5. Runtime Performance

````python

# KPI 5: Cast & Crew Insights

# Filter actors from credits dataset
actors = credits[credits['role'] == 'ACTOR']

# Filter directors from credits dataset
directors = credits[credits['role'] == 'DIRECTOR']

# Count unique actors
total_actors = actors['name'].nunique()

# Count unique directors
total_directors = directors['name'].nunique()

print(" Total Actors:", total_actors)
print(" Total Directors:", total_directors)
````

# 6. Talent KPIs

````python

# KPI 6: Premium Content Analysis

# Filter titles with IMDb and TMDB ratings greater than or equal to 8
premium_titles = titles[
    (titles['imdb_score'] >= 8) &
    (titles['tmdb_score'] >= 8)
]

# Calculate premium content percentage
premium_ratio = (len(premium_titles) / len(titles)) * 100

print(" Premium Titles:", len(premium_titles))
print(f" Premium Content Ratio: {premium_ratio:.2f}%")
````

# 7. Platform Performance Score

````python

# KPI 7: Platform Performance Score

# Calculate the average TMDB popularity score
print(" Average TMDB Popularity:",
      round(titles['tmdb_popularity'].mean(), 2))

# Filter titles having TMDB popularity greater than the median popularity
popular = titles[
    titles['tmdb_popularity'] >
    titles['tmdb_popularity'].median()
]

# Display the total number of popular titles
print(" Popular Titles:",
      len(popular))
````
